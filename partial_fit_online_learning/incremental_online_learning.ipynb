```python
from sklearn import linear_model
```

```python
import numpy as np
```

```python
import time
```

```python
# Generate sample data using numpy
# 1. First understand: What is X and y?
# In Machine Learning, we usually have:
# X (features): Input information given to the model.
# y (target): Correct answer that the model should predict.
n_samples, n_features = 1, 500
y = np.random.randn(n_samples)
x = np.random.randn(n_samples, n_features)

#What does n_samples, n_features = 1, 500 mean?
# n_samples, n_features = 1, 500
# This creates two Python variables:
# n_samples = 1
# n_features = 500


# n_samples  - How many training examples / rows
# n_features - How many input columns per example (means House area1200 sq ft, Bedrooms, Age , Location score)

# There is only 1 row which contains 500 input features


# 3. How does np.random.randn() generate data?
# np.random.randn(...)
# This generates random numbers from a standard normal distribution.
# Simple ga:
# np.random.randn(5)
# Might generate:
# [ 0.42, -1.15, 0.08, 1.73, -0.35 ]
# These are random floating-point numbers.
# Important: These numbers are generated randomly. Every time you run the code, you may get different values.
# For example:
# np.random.randn(5)
# First run:
# [ 0.42, -1.15, 0.08, 1.73, -0.35]
# Second run:
# [-0.21, 0.77, -1.42, 0.15, 0.93]
# The values are not fixed unless you set a random seed.

# 4. Explain this line: y = np.random.randn(n_samples)
# y = np.random.randn(n_samples)
# Since:
# n_samples = 1
# Python executes:
# y = np.random.randn(1)
# This generates one random target value.
# Example:
# y = [0.73]
# Your actual output may be different.



# What does this represent?

# y is the target or answer that the model should learn to predict.
# For this example:
# y = 0.73

# We are saying:

# For the current training example, the correct target value is 0.73.
# Tenglish: y ante answer. Model ki "correct output idi" ani cheptunnam.

# Shape of y
# y.shape

# Output:
# (1,)

# Meaning:
# 1 target value



# 5. Explain this line: x = np.random.randn(n_samples, n_features)
# x = np.random.randn(n_samples, n_features)

# Substitute the values:
# x = np.random.randn(1, 500)
# This generates:
# 1 row
# 500 columns
# So x shape is:
# (1, 500)
# Example with only 5 features
# To make it easy, imagine:

# n_samples = 1
# n_features = 5

# x = np.random.randn(1, 5)

# Possible output:
# x = [[
#     0.42,
#    -1.15,
#     0.08,
#     1.73,
#    -0.35
# ]]

```

```python
print(y)
print(x)
```

```python
# Intialize Online learning model
clf = linear_model.SGDRegressor() #clf - classifier

# Measure training time for single sample
start_time = time.time()
clf.partial_fit(x, y) # input and output

# That means:
# x → One training example with 500 input features.
# y → One target value, the correct answer.

# What does partial_fit() do?
# It gives the sample to SGDRegressor and updates the model's learned parameters.
# Tenglish: Oka sample model ki ichi, daani nunchi nerchukoni weights update cheyyamani cheptunnam.
# For example:
# Input X = [500 feature values]
# Target y = [0.73]

# The model:
# Uses the input to calculate a prediction.
# Compares prediction with the target.
# Calculates the error.
# Updates its weights using SGD.

# This is one incremental training update.

elapsed_time = time.time() - start_time
print("Training time for single sample: ", elapsed_time)
```

```python
y_new = np.random.randn(n_samples)
x_new = np.random.randn(n_samples, n_features)
```

```python
start_time = time.time()
clf.partial_fit(x_new, y_new)
elapsed_time = time.time() - start_time
print("Training time for single sample: ", elapsed_time)
```

### 1. one-hot encoding using the pandas

```python
# creates the dummy variables to represent nominal categories
pd.get_dummies(df, columns=['fuel','owner'])
```

### K-1 OnehotEncoding - keep only N-1 categories

```python
pd.get_dummies(df, columns=['fuel','owner'], drop_first=True)
```

### OnehotEncoding Using Sklearn

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(df.iloc[:,0:4],
    df.iloc[:,-1], test_size=0.2, random_state=2)
```

```python
df.head()
```

```python
X_train, X_test, Y_train, Y_test
```

```python
X_train.head(), X_test.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(drop="first")
### you can use ohe = OneHotEncoder(drop="first", sparse=False) to avoid the
### extra toarray() to convert to numpy array

###array([['Hyundai', 35000, 1.0, ..., 0.0, 0.0, 0.0], if data is float, convert to int below
### ohe = OneHotEncoder(drop="first", sparse=False,dtype=int32)
```

```python
ohe.fit(X_train[['fuel','owner']])
```

```python
X_train_new = ohe.transform(X_train[['fuel','owner']])
X_train_new = X_train_new.toarray()
X_train_new
```

```python
X_test_new = ohe.transform(X_test[['fuel','owner']])
```

```python
X_test_new = X_test_new.toarray()
X_test_new
```

```python
X_train[['brand','km_driven']].values
```

```python
np.hstack((X_train[['brand','km_driven']].values, X_train_new))
```

```python
np.hstack((X_train[['brand','km_driven']].values, X_train_new)).shape
```

```python
### Find whether student gets placed or not using the CGPA and IQ , predict placed(1) and not placed(0)
### Logistic regression is a supervised machine learning algorithm used to predict the probability of a categorical target variable
# How It WorksLinear Equation: 
# 1. It starts by computing a linear combination of the input features,  similar to linear regression (\(z = w^T X + b\)).
# 2. Sigmoid Function: It passes this raw output through a sigmoid function to squash the value into a probability range between 0 and 1.
# 3. Decision Threshold: It applies a threshold (usually 0.5) to convert that probability into a discrete class label (like 0 or 1, Yes or No).

# Types of Logistic Regression
# • Binary: Used when there are only two possible outcomes (e.g., Spam or Not Spam).
# • Multinomial: Used for three or more categories with no specific order (e.g., cat, dog, or sheep).
# • Ordinal: Used for three or more categories that have a natural ranking or order (e.g., low, medium, or high)

# Common Applications
# • Spam Detection: Filtering emails into inbox or spam folders.
# • Medical Diagnosis: Predicting whether a patient has a specific disease or not.
# • Churn Prediction: Identifying if a customer will cancel a subscription.
# • Finance: Determining if a loan applicant is likely to default.
%pip install kagglehub
```

```python
df = pd.read_csv("exam_scores.csv")
print(df.head())
```

```python
print(df.info())
```

```python
print(df.sample(5)) #random 5 samples based on specificed number, head gives 5 samples from head
print(df.tail()) # 5 samples from the end
```

```python
x_train = df.iloc[:, :-1]
y_train = df.iloc[:, -1]

from sklearn.model_selection import train_test_split
print(x_train, y_train)

x_train_input, x_test_input, y_train_input, y_test_input = train_test_split(x_train, y_train, test_size=0.2)
print(x_train_input)
print(x_test_input)

from sklearn.preprocessing import StandardScaler

model = StandardScaler()
model.fit_transform(x_train_input)
model.transform(x_test_input)

from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(x_train_input, y_train_input)
prediction_result = model.predict(x_test_input)

from sklearn.metrics import accuracy_score, mean_squared_error, r2_score

score = accuracy_score(y_test_input, prediction_result)
print("Accuracy score", score)

score_mse = mean_squared_error(y_test_input, prediction_result)
print("Mean Squared error", score_mse)

score_r2score = r2_score(y_test_input, prediction_result)
print("r2_score", score_r2score)


# 1. Accuracy Score: 0.925 (92.5%)
# • What it means: Your model guessed the exact correct label 92.5% of the time on your test dataset. Out of 1,000 test samples, it got 925 right and missed 75.
# • Context: For a classification problem, this is a very strong score.
# 2. Mean Squared Error (MSE): 0.075
# • What it means: This calculates the average squared error of your predictions. 
# Because your targets are likely 0 and 1 (binary classification), a squared error for a wrong guess is still just \(1\) (\(1^2 = 1\)).
# • The Math: Notice that \(1 - 0.925 = 0.075\). Because this is classification, your MSE is simply showing the percentage of incorrect guesses (7.5% wrong). While mathematically true here, MSE is generally ignored for classification tasks.
# 3. \(R^{2}\) Score: 0.6428 (64.3%)
# • What it means: In a traditional regression setup, this would mean your model explains 64.3% of the variance in the data. However, you should ignore this value for your current model.
# • Why it's misleading here: You are applying a regression metric to a classification problem. The math works out to roughly 0.64, but it does not represent the actual real-world performance of your classifier.
```

```python
from sklearn.model_selection import cross_val_score


# Calculate the cross-validation scores
# 'cv=5' means 5-fold cross-validation. 
# 'scoring' defines what metric to use ('accuracy', 'f1', etc.)
cv_scores = cross_val_score(model, x_train_input, y_train_input, cv=5, scoring='accuracy')

# 3. Print the results
print("Scores for each fold:", cv_scores)
print(f"Average Accuracy: {cv_scores.mean():.3f}")
print(f"Standard Deviation (Consistency): {cv_scores.std():.3f}")
```
